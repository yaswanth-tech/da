```python
import matplotlib.pyplot as plt
products = ['Laptop', 'Mobile', 'Tablet', 'Headphones', 'Smartwatch']
sales = [450, 700, 300, 250, 200]

plt.bar(products, sales)
plt.title('Product Sales Comparison')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
products = ['Laptop', 'Mobile', 'Tablet', 'Headphones', 'Smartwatch']
sales = [450, 700, 300, 250, 200]

highest = products[sales.index(max(sales))]
lowest = products[sales.index(min(sales))]

print("Highest sales:", highest)
print("Lowest sales:", lowest)

plt.bar(products, sales)
plt.title('Highest and Lowest Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
departments = ['HR', 'IT', 'Sales', 'Finance', 'Marketing']
employees = [20, 45, 60, 30, 35]

plt.bar(departments, employees)
plt.title('Employees by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.show()
```

```python
departments = ['HR', 'IT', 'Sales', 'Finance', 'Marketing']
employees = [20, 45, 60, 30, 35]

highest_department = departments[employees.index(max(employees))]

print("Department with highest employees:", highest_department)
```

```python
categories = ['Electronics', 'Clothing', 'Furniture', 'Grocery']
sales = [5000, 3000, 2500, 1500]

plt.pie(sales, labels=categories, autopct='%1.1f%%')
plt.title('Sales Distribution by Category')
plt.show()
```

```python
age_groups = ['18-25', '26-35', '36-45', '46-55', '56+']
customers = [120, 180, 150, 100, 50]

plt.bar(age_groups, customers)
plt.title('Customer Distribution by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.show()
```

```python
products = ['Laptop', 'Mobile', 'Tablet', 'Headphones', 'Smartwatch']
sales = [450, 700, 300, 250, 200]

departments = ['HR', 'IT', 'Sales', 'Finance', 'Marketing']
employees = [20, 45, 60, 30, 35]

# Product sales
plt.bar(products, sales)
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()

# Employee counts
plt.bar(departments, employees)
plt.title('Employee Counts')
plt.xlabel('Departments')
plt.ylabel('Employees')
plt.show()
```

```python
categories = ['Electronics', 'Clothing', 'Furniture', 'Grocery']
sales = [5000, 3000, 2500, 1500]

lowest_category = categories[sales.index(min(sales))]

print("Category with lowest contribution:", lowest_category)

plt.bar(categories, sales)
plt.title('Category Sales Comparison')
plt.xlabel('Categories')
plt.ylabel('Sales')
plt.show()
```
