```python
import matplotlib.pyplot as plt
products = ['Laptop', 'Mobile', 'Tablet', 'Headphones', 'Smartwatch']
sales = [450, 700, 300, 250, 200]

plt.bar(products, sales)
plt.title('Product Sales Comparison')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
products = ['Laptop', 'Mobile', 'Tablet', 'Headphones', 'Smartwatch']
sales = [450, 700, 300, 250, 200]

highest = products[sales.index(max(sales))]
lowest = products[sales.index(min(sales))]

print("Highest sales:", highest)
print("Lowest sales:", lowest)

plt.bar(products, sales)
plt.title('Highest and Lowest Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
products = ['Laptop', 'Mobile', 'Tablet', 'Headphones', 'Smartwatch']
sales = [450, 700, 300, 250, 200]

departments = ['HR', 'IT', 'Sales', 'Finance', 'Marketing']
employees = [20, 45, 60, 30, 35]

# Product sales
plt.bar(products, sales)
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()

# Employee counts
plt.bar(departments, employees)
plt.title('Employee Counts')
plt.xlabel('Departments')
plt.ylabel('Employees')
plt.show()
```

```python
categories = ['Electronics', 'Clothing', 'Furniture', 'Grocery']
sales = [5000, 3000, 2500, 1500]

lowest_category = categories[sales.index(min(sales))]

print("Category with lowest contribution:", lowest_category)

plt.bar(categories, sales)
plt.title('Category Sales Comparison')
plt.xlabel('Categories')
plt.ylabel('Sales')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = X**2 + Y**2

plt.contour(X, Y, Z)
plt.title("Contour Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
