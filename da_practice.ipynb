```python
#day 1 11-08
def isPrime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
#groupby
emp_details = {
    'emp_id' :[71,72,73,74,71],
    'city' : ['hyderabad','chennai','bangalore',None,'hyderabad'],
    'experience' : [None, 4, 3, 5, 2],
    'age':[25,30, None, 35,25]

}
df=pd.DataFrame(emp_details)
print(df)
print("-------------------------")
print(df.groupby('city')['age'].mean())
print("-------------------------")
print(df.groupby('city')['age'].median())
print("-------------------------")
print(df.groupby('city')['age'].min())
print("-------------------------")
print(df.groupby('city')['age'].max())
print("-------------------------")
print(df.groupby('city')['age'].count())
print("-------------------------")
print(df.groupby('city')['age'].agg([    #aggregations on single column
    'mean',
    'max',
    'min',
    'sum'
]))
print(df.groupby('city').agg({            #multiple aggregations on multiple columns
    'age':['mean','max'],
    'experience':['sum','count']
}))
```

```python
#day-6 21-06
#merging and joining
df1 = pd.DataFrame({
    'id':[7,8,9],
    'name':['M','S','D']
})

df2 = pd.DataFrame({
    'id':[7,1,2],
    'name':[77,777,7777]
})

merged = pd.merge(
    df1,
    df2,
    on = 'id',
    how = 'outer'    #it joins the both tables rows
)
print(merged)
print("-------------------------")
merged = pd.merge(
    df1,
    df2,
    on = 'id',
    how = 'inner'    #it joins only common rows
)
print(merged)
print("-------------------------")
merged = pd.merge(
    df1,
    df2,
    on = 'id',
    how = 'left'     #it joins the left table rows and common rows
)
print(merged)
print("-------------------------")
merged = pd.merge(
    df1,
    df2,
    on = 'id',
    how = 'right'    #it joind right table rows and common rows
)
print(merged)
print("-------------------------")

#concat

combined = pd.concat(
    [df1,df2],
    axis = 0,            #adds rows vertically
    ignore_index=True    #re arranges
)
print(combined)
print("-------------------------")
combined = pd.concat(
    [df1,df2],
    axis = 1,            #adds columns hortizonatally
)
print(combined)
```

```python
#duplicates
import pandas as pd
data = {
    "EmpID":[101,102,103,104,105,106,107,108,109,110,110],
    "Name":["Ravi","PRIYA","Arun","sneha","Kiran","ANJALI","Vikram","meena","RAHUL","Sneha","Sneha"],
    "City":["Hyderabad","Hyderabad","CHENNAI","Chennai","Bangalore","BANGALORE","Mumbai","mumbai","Delhi","Hyderabad","Hyderabad"],
    "Salary":[30000,40000,35000,45000,32000,50000,38000,42000,36000,45000,45000]
}
df=pd.DataFrame(data)
print(df)
print("____________________")
print(df.duplicated())       #finds duplicates, and if duplicate returns true
print("____________________")
print(df.duplicated().sum()) #total number of duplicate rows
print("____________________")
duplicate = df[df.duplicated()]
print(duplicate)             #display duplicate rows
print("____________________")
duplicate = df[df.duplicated(keep=False)]
print(duplicate)             #display actual and duplicate rows
print("____________________")
duplicate_rows = df[df.duplicated()]
total = duplicate_rows["Salary"].sum()  #sum of salary of duplicate row
print(total)
print("____________________")
duplicate_rows = df[df.duplicated(keep = False)]
total = duplicate_rows["Salary"].sum()        #sum of salary of actual and duplicate row
print(total)
print("____________________")
print(df['City'].duplicated())                #identify and return true for only selected column
print("____________________")
print(df[df['EmpID'].duplicated()])           #identify and return the duplicated entire row
print("____________________")
print(df[df["EmpID"].duplicated(keep=False)]) #return entire actual and duplicate rows/using primary key easy to find
print("____________________")
df = df.drop_duplicates()                    #removes duplicates, keep = "first"-last duplicate is removed, keep = "last" - first row will be removed
print(df)
print("_________________")
df["City"]=df["City"].str.title()
print(df)
```

```python
#Dataset 1 – Customer Transactions
#This dataset represents customer purchase transactions.

import pandas as pd

data = {

   "TransactionID": [1001, 1002, 1003, 1004, 1005,

                      1006, 1007, 1008, 1009, 1010, 1010, 1011],

   "Customer": ["Ravi", "Priya", "Arun", "Sneha", "Kiran",

                "Anjali", "Vikram", "Meena", "Rahul", "Sneha",

                "Sneha", "Kiran"],

   "Product": ["Laptop", "Mobile", "Tablet", "Headphones", "Mouse",

               "Keyboard", "Monitor", "Laptop", "Mobile", "Headphones",

               "Headphones", "Mouse"],

   "Amount": [55000, 25000, 30000, 5000, 1500,

              2500, 18000, 55000, 22000, 5000,

              5000, 1500]}

df_sales = pd.DataFrame(data)

print(df_sales)
print("____________________")
print(df_sales[df_sales.duplicated()])
print("____________________")
print(df_sales.duplicated().sum())
print("____________________")
print(df_sales[df_sales.duplicated(keep=False)])
print("____________________")
duplicate_rows = df_sales[df_sales.duplicated()]
total = duplicate_rows["Amount"].sum()  #sum of salary of duplicate row
print(total)
print("____________________")
print(df_sales[df_sales['TransactionID'].duplicated()])
print("____________________")
print(df_sales[df_sales['Customer'].duplicated()])
print("____________________")
df_sales = df_sales.drop_duplicates()
print(df_sales)
print("____________________")
print(df_sales.drop_duplicates(subset=["TransactionID"],keep = "first"))        #dropping duplicates of specific column
print("_________________")
print(df_sales.drop_duplicates(subset=["Customer","Product"],keep= "first"))
print("_________________")
```

```python
import pandas as pd
import numpy as np
dates = pd.date_range(start="2025-01-01",periods=60,
    freq="D")
sales = np.random.randint(
    2000,
    10000,
    size=60
)
df = pd.DataFrame({
    "Date": dates,
    "Sales": sales
})

print(df)
print("______________________")
df["Date"] = pd.to_datetime(df["Date"])
print(df)
#year
print("______________________")
df["Date"].dt.year
print(df[df["Date"].dt.year == 2026])
print("______________________")
print(df.groupby(df["Date"].dt.year)["Sales"].sum())
print("______________________")
print(df.groupby(df["Date"].dt.year)["Sales"].max())
print("______________________")
print(df.groupby(df["Date"].dt.year)["Sales"].min())
print("______________________")
print(df.groupby(df["Date"].dt.year)["Sales"].mean())
print("______________________")
print(df.groupby(df["Date"].dt.year)["Sales"].count())
print("______________________")
print(df.groupby(df["Date"].dt.year)["Sales"].median())
print("______________________")
```

```python
#numpy random package
ran = np.random.random(7)           #0-1 float numbers
print(ran)
print("______________")
ran1 = np.random.random((2,3))      #for 2d
print(ran1)
print("______________")
ran3 = np.random.random((2,3,4))    #for 3d
print(ran3)
print("______________")
ran4 = np.random.uniform(7, 77, 7)  #uniform- here(low, high, size)
print(ran4)
print("______________")
ran5 = np.random.uniform(7,77,(2,4)) #2d
print(ran5)
print("______________")
ran6 = np.random.uniform(7,77,(2,2,3))#3d
print(ran6)
print("______________")
ran7 = np.random.randn(7)            #randn- gives random values with mean = 0, standard deviation = 1
print(ran7)
print(ran7.mean())
print(ran7.std)
print("______________")
ran7 = np.random.randn(2,3)            #randn- 2d
print(ran7)
print(ran7.mean())
print(ran7.std)
print("______________")
ran7 = np.random.randn(2,2,3)            #randn-3d
print(ran7.mean())
print(ran7.std)
print("______________")
cho = np.array([7,77,777,7777])         #choice- gives the random values from the given array
result = np.random.choice(cho)          #gives multiple use (arr,number)
print(result)
print("______________")
str = ["yash","kams","hasi","raju","raj"]  #gives with string
result = np.random.choice(str,2)
print(result)
print("______________")
pro = [7,77,777]                           #choice probability, here we have, arr, size = number, p = sum should be 1
result = np.random.choice(
    pro,
    size = 4,
    p=[0.5,0.2,0.3]
)
print(result)
print("______________")
sd = np.random.uniform(7, 77)
np.random.seed(10)                        #seed - freeze the random number, so it doesn't change, it doesn't generate, seed(nummber) - number is number of times it freezes
print(sd)
print("______________")
sd1 = np.random.random(7)
np.random.seed(7)
print(sd1)
print("______________")
sd2 = np.random.randn(7)
np.random.seed(7)
print(sd2)
print("______________")
cho = np.array([7,77,777,7777])
result = np.random.choice(cho)
np.random.seed(7)
print(result)
print("______________")
cho = np.array([7,77,777,7777])        #shuffle- different orders  - changes original array
np.random.shuffle(cho)
print(cho)
print("______________")
cho = np.array([7,77,777,7777])       #permutation-orginal will be same, permutation will duplicate order
result = np.random.permutation(cho)
print("Original":cho)
print("Permutation":result)
print("______________")
```

```python
#day-13 29-08
#linspace-- its the difference equally b/w numbers
import numpy as np
lin = np.linspace(0, 7, 7)
print(lin)
print("_____________________")
#shape- attribute for giving rows and columns
arr = np.array([10,20,30,40])
print(arr.shape)
print("_____________________")
#ravel - compress multi into single(1-d)

ra = np.array([[1,2,3,4],[5,6,7,8]]) #2d to 1d
result = ra.ravel()
print(result)

ra1 = np.arange(1,10).reshape(3,3,1)   #3D TO 1D
result = ra1.ravel()
print(result)
print("_____________________")
#transponse
tra = np.arange(1,11).reshape(2,5)
print(tra)
r = tra.T
print(r)
print("_____________________")

#Data type conversion
dt = np.array([10,20,30,40])
print(dt)
print(dt.dtype)
data = dt.astype(float)
print(data)
print(data.dtype)
print("_____________________")
arr = np.array([10,20,30,40,50])    #1d
print(arr[-3:-1])     #last 2rd, 3rd
print(arr[-2:])       #last 2
print(arr[0:2])       #first 2

arr1 = np.arange(1,11).reshape(2,5)   #2d
print(arr1)
print(arr1[0,3])
print(arr1[0:1])        #slicing rows
print(arr1[0:2,0:4])     #slicing in 2d for rows and columns
print("_____________________")
#modification
mod = np.array([10,20,30,40,50])   #1d
mod[2] = 70
print(mod)

mod1 = np.arange(1,11).reshape(2,5) #2d
mod1[0] = 7
print(mod1)

mod[1:3] = 7   #multiple elements in 1d
print(mod)
print("_____________________")
#insert
ns = np.insert(mod, 4, 77)
print(ns)

ns = np.insert(mod, 3, [7,77])   #inserting multiple values
print(ns)
print("_____________________")
#delete
de = np.array([10,20,30,40,50])
dele = np.delete(de, 4)
print(dele)

dele = np.delete(de, [1,2])   #deleting multiple values
print(dele)

```
