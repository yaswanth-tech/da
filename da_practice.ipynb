```python
#day 1 11-08
def isPrime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
#filtering in df
data = {
    'name':['yaswanth','lakshman','nandan','varun','sami'],
    'age':[21,25,None,23,24],
    'city':['vizag','hyderabad','bombay',None, 'bengaluru'],
    'salary':[77700,7700,7777,777,None]
}

i = [71, 72, 73, 74, 75]

df = pd.DataFrame(data, index = i)
print(df)
print("-------------------------")
print(df[df['age']>21])  #df[df] gives values of particular condition, wheres as df will give boolean values
print("-------------------------")
print(df[df['salary']>40000])
print("-------------------------")
print(df[df['city']=='hyderabad'])
print("-------------------------")
print(df[(df['age'] > 22) & (df['city'] == 'hyderabad')])
print("-------------------------")
print(df[(df['city']=='bombay') | (df['salary'] > 25000)])
print("-------------------------")
print(df.isna())       # returns the boolean value, returns true if nan, return false if value
print("-------------------------")
print(df.isna().sum()) # count of missing values
print("-------------------------")
print(df.fillna(0))    # fill the value u want to replace
print("-------------------------")
print(df.dropna())     # drop the missing values
print("-------------------------")
print(df.dropna(axis=1))  #removing columns missing values
print("-------------------------")
print(df.dropna(axis=0))
print("-------------------------")
print(df.dropna(how='all'))
print("-------------------------")
print(df.dropna(inplace=True))
print("-------------------------")
print(df.fillna(0))
print("-------------------------")
b = df.fillna({              #filling different columns with different values
    'age': 0,
    'city': 'Unknown'
})
print(b)
print("-------------------------")
df['age'] = df['age'].fillna(df['age'].mean())  #fill numerical missing values with mean
```

```python
emp_details = {
    'emp_id' :[71,72,73,74,71],
    'city' : ['hyderabad','chennai','bangalore',None,'hyderabad'],
    'experience' : [None, 4, 3, 5, 2],
    'age':[25,30, None, 35,25]

}
df=pd.DataFrame(emp_details)
print(df)
print("-------------------------")
print(df.isna())
print("-------------------------")
print(df.dropna())
print("-------------------------")
print(df.isna().sum())
print("-------------------------")
b = df.fillna({           #experience =0, city = unknown
    'experience': 0,
    'city': 'Unknown'
},inplace = True)
print(b)
df['age'] == df['age'].fillna(df['age'].median(), inplace = True)   #age=median
print(df)
print("-------------------------")
df.rename(columns={"emp_id":'employee_id'},inplace=True)   #replace emp_id as employee_id
#create new column as salary for all employees --- manipulating
df['salary'] = [77000, None, 7777, 77770, 777]
print(df)
print("-------------------------")
df['age'] = df['age'] + 1        #age + 1
print(df)
print("-------------------------")
df['bonus'] = df['salary'] / 10  #10% bonus to salary
print(df)
print("-------------------------")
df.drop('bonus', axis =1 , inplace = True)  #drop bonus column
print(df)
print("-------------------------")
#df.drop(columns=['age', 'salary'],inplace=True)-drop multiple columns
df.loc[len(df)] = [77, 'vizag', 7, 17, 77]
print(df)
print("-------------------------")
df.drop(2,axis = 0, inplace = True)   #dropping a specific row
print(df)
print("-------------------------")
```
