```python
#day-17 04-09
#matplotlib
import matplotlib.pyplot as plt
import numpy as np

products = ["laptop","mobile","tablet","headphones","watch"]
sales = [120, 180, 90, 150,110]
profit = [30,50,20,40,25]
```

```python
#line graph
plt.plot(products, sales, marker = "o")
plt.title("product vs sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python
#multi-line graph
plt.plot(products, sales, marker = "o", label = "sales")
plt.plot(products, profit , marker = "*",label = "products")
plt.title("sales vs profit")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()
plt.show()
```

```python
#bar-graph
plt.bar(products, sales, label = "sales")
plt.title("products vs sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.legend()
plt.show()
```

```python
#horizontal bar graph
width = 0.35
plt.barh(products, sales, width, label = "sales")
plt.title("products vs sales")
plt.xlabel("sales")
plt.ylabel("product")
plt.legend()
plt.show()
```

```python
#clustered bar grpah
x = np.arange(len(products))
width = 0.35
plt.bar(x-width/2,sales, width, label= "sales")
plt.bar(x+width/2,profit,width,label ="profit")
plt.title("sales vs profit")
plt.xlabel("product")
plt.ylabel("sales")
plt.legend()
plt.show()
```

```python
#stacked bar grpah
x = np.arange(len(products))
width = 0.35
plt.bar(x+width/2,sales, width, label= "sales")
plt.bar(x+width/2,profit,width,label ="profit")
plt.title("sales vs profit")
plt.xlabel("product")
plt.ylabel("sales")
plt.legend()
plt.show()
```

```python
#day -18 08-09
#line chart
month = ["jan","feb","mar","apr","may","june"]
monthly_sales = [100,120,150,140,180,200]
plt.plot(month, monthly_sales, marker = "o")
plt.title("monthly sales")
plt.xlabel("month")
plt.ylabel("sales")
plt.show()
```

```python
#area chart
plt.plot(month, monthly_sales)
plt.fill_between(month, monthly_sales)
plt.title("monthly sales")
plt.xlabel("month")
plt.ylabel("sales")
plt.show()
```

```python
#pie chart
plt.pie(
    sales,
    labels = products,
    autopct="%1.1f%%"
)

plt.title("product sales distribution")
plt.show()
```

```python
#Donut graph
import matplotlib.pyplot as plt

products = ["laptop", "mobile", "tablet", "watch"]
sales = [120, 180, 90, 110]

plt.pie(
    sales,
    labels=products,
    wedgeprops={"width": 0.3}
)

plt.text(
    0, 0,
    "Total 500",
    ha="center",
    va="center"
)

plt.show()
```

```python
#day-20 10-09
import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Math": [65, 70, 75, 80, 85, 60, 90, 72, 78, 88],
    "Python": [70, 75, 80, 85, 90, 65, 95, 75, 82, 92],
    "SQL": [60, 68, 72, 78, 82, 55, 88, 70, 75, 85]
}

df = pd.DataFrame(data)

print(df)
```

```python
#violin plot
import matplotlib.pyplot as plt
plt.violinplot(df["Math"])
plt.title("Distribution of Student Marks")
plt.ylabel("Marks")
```

```python
#stem plot
plt.stem(df["Student"], df["Math"])
plt.title("Student Math Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
#contour plot
import numpy as np
import matplotlib.pyplot as plt

# x = np.arange(1, 11)
# y = np.arange(1, 11)

x = [1, 2, 3]
y = [10, 20]

X, Y = np.meshgrid(x, y)

Z = X**2 + Y**2

plt.contour(X, Y, Z)
plt.title("Contour Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
#heatmap
marks = df[["Math", "Python", "SQL"]].values

plt.imshow(marks)
plt.title("Student Marks Heatmap")
plt.xlabel("Subjects")
plt.ylabel("Students")
plt.colorbar()
```
