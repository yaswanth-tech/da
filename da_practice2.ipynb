```python
import pandas as pd

data = {
    'name':['Ravi','Priya','Arun','Sneha','Kiran','Anjali','Vikram','Menna'],
    'age':[22,24,21,25,23,26,20,27],
    'city':["hyderabad","chennai","bangalore","hyderabad","mumbai","chennai","bangalore","hyderabad"],
    'salary':[30000,40000,28000,50000,35000,45000,25000,55000]
}

df = pd.DataFrame(data)
print(df)
print("____________________")
print(df["name"])
print("____________________")
print(df[["name","salary"]])
print("____________________")
print(df[df["age"]>23])
print("____________________")
print(df[df["salary"]>40000])
print("____________________")
print(df[df["city"]=="chennai"])
print("____________________")
print(df[(df["age"] > 22) & (df["salary"] > 35000)])
print("____________________")
print(df[(df["city"] == "mumbai") | (df["city"] == "bangalore")])
print("____________________")
print(df[df["salary"] > 30000][["name", "salary"]])
print("____________________")
print(df.loc[1:3]) #we can use iloc[1:4]
print("____________________")
print(df.iloc[:4][["name", "age"]])  #we acan also use loc[:3]
print("____________________")
print(df.iloc[[0,3,6]])
print("____________________")
print(df.loc[2:5])
print("____________________")
df = df.rename(columns={
    "name": "Employee_Name",
    "age": "Employee_Age",
    "city": "Location",
    "salary": "Monthly_Salary"
})
print(df)
print("____________________")
print(df.loc[
    (df["Employee_Age"] >= 22) & (df["Employee_Age"] <= 26),
    ["Employee_Name", "Location", "Monthly_Salary"]
])
```
